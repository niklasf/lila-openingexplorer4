```python
print("i", "distinct", "1 occ", "2 occ", "3 occ", "4 occ", "5 occ", "max occ", sep=";")
for i in range(1, 30):
    with open(f"lichess_db_{i}.txt") as f:
        max_occs = None
        by_occs = [0, 0, 0, 0, 0]
        distinct = 0
        for line in f:
            if "occurences: " in line:
                occs, _, pos, _ = line.split(None, 3)
                occs = int(occs)
                pos = int(pos)
                if max_occs is None:
                    max_occs = occs
                if 1 <= occs <= 5:
                    by_occs[occs - 1] = pos
            if "distinct positions" in line:
                d, _ = line.split(None, 1)
                d = int(d)
                distinct = d
        print(i, distinct, ";".join(str(p) for p in by_occs), max_occs, sep=";")
```
